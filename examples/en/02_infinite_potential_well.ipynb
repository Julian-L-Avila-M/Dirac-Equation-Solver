# Tutorial 02: Relativistic Infinite Potential Well (1D & 2D)

> **Language / Idioma:** [Español](../es/02_pozo_potencial_infinito.ipynb) | **English**

This notebook simulates a relativistic Dirac particle confined inside an `InfiniteWellPotential`:
$$V(\mathbf{r}) = \begin{cases} 0 & \text{if } |r_k| < L_k / 2 \;\forall k \\ \infty & \text{otherwise} \end{cases}$$

```python
import sys
sys.path.append('../..')
import numpy as np
import matplotlib.pyplot as plt
from dirac_solver import (
    Grid, ConstantSpinor, GaussianPacket,
    DiracProblemBuilder, DiracSolver, InfiniteWellPotential, Observables, _core
)
```

## 1. 1D Confinement Inside an Infinite Well

```python
grid_1d = Grid(shape=(200,), spacing=(0.1,))
spinor = ConstantSpinor(momentum=[3.0, 0.0, 0.0], mass=1.0)
packet = GaussianPacket(spinor, center=[0.0, 0.0, 0.0], spatial_width=1.0)
well_1d = InfiniteWellPotential(widths=[12.0])

problem_1d = (
    DiracProblemBuilder()
    .set_grid(grid_1d)
    .set_initial_state(packet)
    .set_potential(well_1d)
    .set_boundary_condition(_core.AbsorbingBoundary(strength=0.1, width=10))
    .set_time_parameters(time_step=0.002, total_time=4.0)
    .build()
)

solver_1d = DiracSolver(problem_1d)
obs_1d = Observables(solver_1d)
solver_1d.run_simulation(observables_handler=obs_1d)
solver_1d.plot_probability_density()

times = obs_1d.get_history('time')
pos_x = obs_1d.get_history('expected_position')[:, 0]
plt.figure(figsize=(9, 4))
plt.plot(times, pos_x)
plt.xlabel('Time $t$')
plt.ylabel('Expected position $\\langle x \\rangle$')
plt.title('1D Oscillatory Motion Inside an Infinite Potential Well')
plt.grid(True)
plt.show()
```

## 2. 2D Box Confinement

```python
grid_2d = Grid(shape=(80, 80), spacing=(0.1, 0.1))
spinor_2d = ConstantSpinor(momentum=[2.5, 1.5, 0.0], mass=1.0)
packet_2d = GaussianPacket(spinor_2d, center=[0.0, 0.0, 0.0], spatial_width=0.8)
well_2d = InfiniteWellPotential(widths=[6.0, 6.0])

problem_2d = (
    DiracProblemBuilder()
    .set_grid(grid_2d)
    .set_initial_state(packet_2d)
    .set_potential(well_2d)
    .set_boundary_condition(_core.AbsorbingBoundary(strength=0.1, width=5))
    .set_time_parameters(time_step=0.002, total_time=2.5)
    .build()
)

solver_2d = DiracSolver(problem_2d)
solver_2d.run_simulation()
solver_2d.plot_probability_density()
```
