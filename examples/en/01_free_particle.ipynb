In this tutorial you will learn how to simulate a free Dirac fermion ($V(\mathbf{r}) = 0$) in 1D, 2D, and 3D using `GaussianPacket` and `PlaneWave` initial states with both `PeriodicBoundary` and `AbsorbingBoundary` conditions.

The free Dirac equation in natural units ($\hbar = c = 1$) is:
$$i \frac{\partial \psi}{\partial t} = (-i\boldsymbol{\alpha} \cdot \nabla + m\beta)\psi$$

```python
import sys
sys.path.append('../..')
import numpy as np
import matplotlib.pyplot as plt
from dirac_solver import (
    Grid, ConstantSpinor, GaussianPacket, PlaneWave,
    DiracProblemBuilder, DiracSolver, FreeParticle, Observables, _core, electron_mass
)
```

## 1. 1D Free Gaussian Wavepacket (Periodic vs. Absorbing Boundaries)

We first check the Courant-Friedrichs-Lewy (CFL) stability limit $\Delta t \le \Delta t_{\max}$ using `grid.courant_limit()`.

```python
grid_1d = Grid(shape=(200,), spacing=(0.1,))
print(f'1D Grid Courant limit dt_max = {grid_1d.courant_limit():.4f}')

spinor_1d = ConstantSpinor(
    momentum=[4.0, 0.0, 0.0],
    spin_polarization=[0, 0, 1],
    spin_superposition=[1 + 0j, 0 + 0j],
    mass=electron_mass,
)
packet_1d = GaussianPacket(constant_spinor=spinor_1d, center=[-3.0, 0.0, 0.0], spatial_width=1.0)

problem_periodic = (
    DiracProblemBuilder()
    .set_grid(grid_1d)
    .set_initial_state(packet_1d)
    .set_potential(FreeParticle())
    .set_boundary_condition(_core.PeriodicBoundary())
    .set_time_parameters(time_step=0.002, total_time=4.0)
    .build()
)

solver_p = DiracSolver(problem_periodic)
obs_p = Observables(solver_p)
solver_p.run_simulation(observables_handler=obs_p)
solver_p.plot_probability_density()
```

### Comparing with Absorbing Boundary Conditions (`AbsorbingBoundary`)

```python
problem_absorbing = (
    DiracProblemBuilder()
    .set_grid(grid_1d)
    .set_initial_state(packet_1d)
    .set_potential(FreeParticle())
    .set_boundary_condition(_core.AbsorbingBoundary(strength=0.25, width=25))
    .set_time_parameters(time_step=0.002, total_time=4.0)
    .build()
)

solver_a = DiracSolver(problem_absorbing)
solver_a.run_simulation()
solver_a.plot_probability_density()
```

## 2. 2D Free Wavepacket Propagation

```python
grid_2d = Grid(shape=(80, 80), spacing=(0.1, 0.1))
print(f'2D Grid Courant limit dt_max = {grid_2d.courant_limit():.4f}')

spinor_2d = ConstantSpinor(momentum=[3.0, 2.0, 0.0], mass=electron_mass)
packet_2d = GaussianPacket(spinor_2d, center=[-1.5, -1.0, 0.0], spatial_width=0.9)

problem_2d = (
    DiracProblemBuilder()
    .set_grid(grid_2d)
    .set_initial_state(packet_2d)
    .set_potential(FreeParticle())
    .set_boundary_condition(_core.PeriodicBoundary())
    .set_time_parameters(time_step=0.002, total_time=2.0)
    .build()
)

solver_2d = DiracSolver(problem_2d)
solver_2d.run_simulation()
solver_2d.plot_probability_density()
```
